```python
from google.colab import drive
import os, time

drive.mount('/content/drive')
cartella = '/content/drive/MyDrive/Collatz_Ricerca'
if not os.path.exists(cartella): os.makedirs(cartella)

FILE_REGISTRO = os.path.join(cartella, "registro_collatz_colab.txt")
START_ESP, MAX_ESP = 10000, 20000
prossimo_p = START_ESP

if os.path.exists(FILE_REGISTRO):
    with open(FILE_REGISTRO, "r") as f:
        righe = f.readlines()
        if len(righe) > 1:
            try: prossimo_p = int(righe[-1].split(";")[0]) + 1
            except ValueError: pass
else:
    with open(FILE_REGISTRO, "w") as f: f.write("Esponente;Cifre;Passaggi;Tempo(s)\n")

print(f"Inizio calcolo dall'esponente: {prossimo_p}")

try:
    while prossimo_p <= MAX_ESP:
        n = (1 << prossimo_p) - 1
        start_val = n
        cifre_stimate = int(prossimo_p * 0.30103) + 1
        passi = 0
        start_time = time.time()

        while n > 1:
            if n % 2 == 0: n = n >> 1
            else: n = 3 * n + 1
            passi += 1
            if n < start_val: break

        tempo_impiegato = time.time() - start_time
        with open(FILE_REGISTRO, "a") as f:
            f.write(f"{prossimo_p};{cifre_stimate};{passi};{tempo_impiegato:.4f}\n")

        print(f"p={prossimo_p} ({cifre_stimate} cifre) -> OK in {passi} passi ({tempo_impiegato:.2f}s)")
        prossimo_p += 1
except KeyboardInterrupt:
    print("\nFermato dall'utente.")
```

```python
import os

# Percorso del file su Google Drive
FILE_REGISTRO = '/content/drive/MyDrive/Collatz_Ricerca/registro_collatz_colab.txt'

if os.path.exists(FILE_REGISTRO):
    with open(FILE_REGISTRO, "r") as f:
        righe = f.readlines()

    # Escludiamo l'intestazione del file
    dati = righe[1:]
    totale_verificati = len(dati)

    if totale_verificati > 0:
        primo_p = dati[0].split(";")[0]
        ultimo_p = dati[-1].split(";")[0]

        # Troviamo il numero che ha richiesto più passaggi
        piu_lento_p = 0
        max_passi = 0
        for riga in dati:
            parti = riga.strip().split(";")
            if len(parti) >= 3:
                p_corrente = int(parti[0])
                passi_correnti = int(parti[2])
                if passi_correnti > max_passi:
                    max_passi = passi_correnti
                    piu_lento_p = p_corrente

        print("📊 === RESOCONTO DELLA TUA RICERCA ===")
        print(f"✅ Numeri totali verificati finora: {totale_verificati}")
        print(f"🚀 Intervallo esponenti controllato: da p={primo_p} a p={ultimo_p}")
        print(f"🔥 L'esponente più 'combattivo' trovato: p={piu_lento_p} con ben {max_passi} passaggi prima di scendere!")
        print("=======================================")
    else:
        print("Il file esiste ma è ancora vuoto. Attendi il completamento del primo esponente!")
else:
    print("❌ Non trovo ancora il file su Google Drive. Assicurati che lo script principale abbia completato almeno un numero!")
```
